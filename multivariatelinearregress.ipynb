```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn import linear_model 
from sklearn.model_selection import train_test_split
```

```python
Z = []
N = []
LINEARCOEFF = []
INTERCEPT = []
for x in "abcdefghijklm":
    df = pd.read_csv(f"test{x}.csv", header=0)
    X = df[["NKEEP"]].to_numpy()
    y = df[["ITERATION"]].to_numpy()
    model = LinearRegression()
    model.fit(X, y)   
    Z.append(df.loc[2,"Z"])
    N.append(df.loc[2,"N"])
    LINEARCOEFF.append(model.coef_)
    INTERCEPT.append(model.intercept_)
```

```python
storage = {"Z": Z, "N": N, "LINEARCOEFF": LINEARCOEFF, "INTERCEPT": INTERCEPT}
equations = pd.DataFrame(data=storage)
equations.to_csv("equationscheck.csv", sep = ',', header = False, index = False)
X=equations[["Z", "N"]]
y=equations[["LINEARCOEFF", "INTERCEPT"]]
print(X)
print(y)
model.fit(X,y)
print(model.predict([[2,2]]))
```
